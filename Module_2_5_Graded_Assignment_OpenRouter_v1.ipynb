### Writing your OpenRouter code for the OpenAI model


### Please Write your answer here

Delete the `raise NotImplementedError()` when you start writing your code.
Use OpenRouter and the `OPENAI_MODEL` model variable.


````python
# --- Student TODO: implement only this function. Keep it minimal. ---
from openai import OpenAI

def call_openai(prompt: str) -> str:
    """
    Return the plain text output from the selected OpenAI model through OpenRouter.
    Use OPENROUTER_API_KEY, OPENROUTER_BASE_URL, and OPENAI_MODEL.
    """
    client = OpenAI(
        api_key=OPENROUTER_API_KEY,
        base_url=OPENROUTER_BASE_URL,
    )
    response = client.chat.completions.create(
        model=OPENAI_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=250,
    )
    return response.choices[0].message.content or ""
````

## What we will test (assert cells)
We will call **your three functions** with three prompts and check shape/format only:

**Prompts we will use:**
```python
CREATIVE_PROMPT = "Write a 4-line poem about learning by testing code."
ANALYTICAL_PROMPT = (
    "Summarize in one sentence (<= 40 words): "
    "'Using multiple LLM providers helps compare creativity, formatting reliability, and speed, "
    "but adds integration overhead and testing complexity.'"
)
INSTRUCTION_PROMPT = (
    "Return ONLY valid JSON (no backticks). The JSON must be an object with key 'todos' "
    "which is a list of EXACTLY 2 items. Each item must be an object with keys 'task' (string) "
    "and 'done' (boolean)."
)
```

**Checks:**
- Creative: output has **≥ 4 non-empty lines**.
- Analytical: output is **one sentence** and **≤ 40 words**.
- Instruction: output is **valid JSON** with exactly **2** todos, each `{ "task": str, "done": bool }`.

That’s all. Keep it simple.


````python
import time, json
import pandas as pd

# Models to compare (same OpenRouter key, different model IDs)
MODELS = {"OpenAI": OPENAI_MODEL, "Gemini": GEMINI_MODEL, "Claude": CLAUDE_MODEL}

# Identical prompts for every model so the model is the only variable
PROMPTS = {
    "Creative": "Write a 4-line poem about learning by testing code.",
    "Analytical": ("Summarize in one sentence (<= 40 words): "
                   "'Using multiple LLM providers helps compare creativity, formatting reliability, and speed, "
                   "but adds integration overhead and testing complexity.'"),
    "Instruction": ("Return ONLY valid JSON (no backticks). The JSON must be an object with key 'todos' "
                    "which is a list of EXACTLY 2 items. Each item must be an object with keys 'task' (string) "
                    "and 'done' (boolean)."),
}

def run_model(model_id, prompt, max_tokens=400):
    """Call one model via OpenRouter; return text, latency and token usage."""
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)
    start = time.time()
    try:
        r = client.chat.completions.create(
            model=model_id,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.3,
            max_tokens=max_tokens,
        )
        return {"text": r.choices[0].message.content or "",
                "seconds": round(time.time() - start, 2),
                "in_tokens": r.usage.prompt_tokens,
                "out_tokens": r.usage.completion_tokens,
                "error": None}
    except Exception as e:
        return {"text": "", "seconds": round(time.time() - start, 2),
                "in_tokens": None, "out_tokens": None, "error": str(e)}

def check(task, text):
    """Simple automatic format checks for each task."""
    if task == "Creative":
        return len([l for l in text.splitlines() if l.strip()]) >= 4
    if task == "Analytical":
        words = len(text.split())
        return words <= 40 and text.strip()[-1:] in ".!?" and text.count(". ") == 0
    if task == "Instruction":
        try:
            d = json.loads(text)
            t = d["todos"]
            return (len(t) == 2 and all(isinstance(i["task"], str)
                    and isinstance(i["done"], bool) for i in t))
        except Exception:
            return False

rows = []
for task, prompt in PROMPTS.items():
    for name, model_id in MODELS.items():
        res = run_model(model_id, prompt)
        res.update({"task": task, "model": name, "format_ok": check(task, res["text"])})
        rows.append(res)
        print(f"\n=== {task} | {name} ({res['seconds']}s) ===\n{res['text']}\n"
              f"{'ERROR: ' + res['error'] if res['error'] else ''}")

df = pd.DataFrame(rows)[["task", "model", "seconds", "in_tokens", "out_tokens", "format_ok"]]
df
````
